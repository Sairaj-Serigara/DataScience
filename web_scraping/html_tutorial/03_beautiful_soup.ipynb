```python
from bs4 import BeautifulSoup
```

```python
with open("html_pages/page-1.html") as f:
    content=f.read()     #reading content

soup=BeautifulSoup(content,"html.parser")   # creating soup
```

```python
h3s=soup.find_all("h3")
h3s
```

```python
for i in h3s:
    print(i.find("a")['title'])   #title is not inside <h3>, it's inside <a>.
                                   # find <a> inside <h3>
                                   # get title from <a>
                                """ h3
                                      └── a
                                          └── text"""
```

```python
articles=soup.select("article.product_pod")
```

```python
items=[]
for i in articles:
    title=i.find("h3").find("a")['title']
    price=i.select_one("p.price_color").text.split("£")[1]  #select_one returns returns one tag,select returns list and list has no .text
    #print(price)                                      #.split("£")[1] is used to return text after £
    rating_ele=i.select_one("p.star-rating")      #class="star-rating One"---->BeautifulSoup sees:['star-rating', 'One']
    rating=rating_ele['class'][1]
    # print(rating)
    stock=i.select_one("p.instock.availability").text.strip()
    # print(stock)
    
    items.append([title,price,rating,stock])
items
```

```python
import pandas as pd
df=pd.DataFrame(items,columns=["Book","price","rating","stock"])
df
```

```python
df.to_csv("web_scrap.csv",index=False)
```
