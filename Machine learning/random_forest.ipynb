```python
from sklearn.ensemble import RandomForestClassifier


features=[[150,0],
          [170,0],
          [130,1],
          [120,1]]

labels=['apple','apple','orange','orange']
clf=RandomForestClassifier()
clf=clf.fit(features,labels)

prediction=clf.predict([[150,0],[13,1]])
print(f"the predicted fruite is {prediction[0]}")
print(f"the predicted fruite is {prediction[1]}")

```
