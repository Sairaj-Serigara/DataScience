```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features=[[150,0],
          [170,0],
          [130,1],
          [120,1]]

labels=['apple','apple','orange','orange']
clf=DecisionTreeClassifier()
clf=clf.fit(features,labels)

prediction=clf.predict([[150,0],[13,1]])
print(f"the predicted fruite is {prediction[0]}")
print(f"the predicted fruite is {prediction[1]}")
```

```python
features1=[[150,1],
          [20,1],
          [100,1],
          [200,0],
          [180,0],
          [220,0]]
labels1=["cat","cat","cat","dog","dog","dog"]
clf=DecisionTreeClassifier()
clf=clf.fit(features1,labels1)
predicted=clf.predict([[130,1]])
print(f"predicted animal is {predicted[0]}")
```
