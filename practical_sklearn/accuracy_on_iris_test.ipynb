```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

data=pd.read_excel("iris/iris-test.xlsx")
x=data.iloc[:, :-1]
y=data.iloc[:, -1]
model=DecisionTreeClassifier()
clf=model.fit(x.values, y.values)
prediction=clf.predict(x.values)
```

```python
actual=y.values
count=0
for i in range(0, len(actual)):
    if prediction[i]==actual[i]:
        count+= 1
accuracy=(count / (len(actual))) * 100
accuracy
```

```python
data_train=pd.read_excel("iris/iris-train.xlsx")
x_train=data_train.iloc[:,:-1]
y_train=data_train.iloc[:,-1]
clf_train=model.fit(x_train.values,y_train.values)
prediction_train=clf_train.predict(x.values)
```

```python
actual_train=y.values   #y_test class
count_train=0
for i in range(0,len(actual_train)):
    if prediction_train[i]==actual_train[i]:
        count_train+=1
accuracy_train=(count_train/(len(actual_train)))*100
accuracy_train
```
